## Graficar

```python
plt.figure()

for mem in df["Memory"].unique():
    subset = df[df["Memory"] == mem]
    plt.plot(subset["Size"], subset["Cycles_per_byte"], marker='o', label=mem)

plt.xlabel("Tamaño de memoria")
plt.ylabel("Ciclos por byte")
plt.title("Tamaño vs Ciclos por Byte por Tipo de Memoria")
plt.legend()
plt.grid()

plt.show()
```
